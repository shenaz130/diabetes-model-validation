```python
import numpy as np
import pandas as pd
import lightgbm as lgb
from scipy.stats import ttest_rel
from sklearn.model_selection import GroupShuffleSplit, StratifiedGroupKFold
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (f1_score, roc_auc_score, recall_score, balanced_accuracy_score,
                             classification_report, confusion_matrix, ConfusionMatrixDisplay)
from imblearn.over_sampling import SMOTENC
import matplotlib.pyplot as plt
```

# Model Development Design
1. The team will use 2 models. Multinomial Logistic Regression (which will be our baseline) and LightGBM. For each model, we will compare weight balanced and SMOTENC balanced.
2. To baseline our dataset, we will have 3 tiers. We will rank the features by:
*   Tier 1 = Top 20 Correlated Features
*   Tier 2 = Tier 1 + discharge_disposition, admission_source, admission_type, diag_1/2/3_category
* Tier 3 = all 110 columns
3. To avoid leakage, we will **rank the features and create the cross validation folds using the training dataset only.** The hold out test set will be **untouched**.
4. Once all tiers and model validation are completed, we will select the best model to apply to the test set in the final model validation.


```python
# Function to rank the features according to correlation within the training set
def rank_features(X, y):
    """One-vs-rest Pearson correlation of each feature with each readmission class."""
    y = pd.Series(y, index=X.index)
    class_labels = {0: "r_NO", 1: "r_<30", 2: "r_>30"}
    ranking = pd.DataFrame(index=X.columns)
    for label, name in class_labels.items():
        ranking[name] = X.corrwith((y == label).astype(int))
    ranking["max_abs_r"] = ranking[list(class_labels.values())].abs().max(axis=1)
    ranking = ranking.dropna().sort_values("max_abs_r", ascending=False)
    ranking.insert(0, "rank", range(1, len(ranking) + 1))
    return ranking

SEED = 42
N_FOLDS = 5
CLASS_NAMES = ["NO", "<30", ">30"]
```

```python
# Drop columns that are not needed or part of the target variable
drop_cols = ["encounter_id", "patient_nbr", "admission_type_id", "discharge_disposition_id",
             "admission_source_id", "diag_1", "diag_2", "diag_3",
             "readmitted", "readmitted_binary", "readmitted_multiclass"]

# Columns to one-hot encode (the same columns the EDA notebook encoded:
# text categories AND the 0/1 medication flags, e.g. glipizide -> glipizide_1)
ENCODE_COLS = ["race", "gender", "age", "admission_type", "discharge_disposition", "admission_source",
               "diag_1_category", "diag_2_category", "diag_3_category",
               "insulin", "metformin", "change", "diabetesMed",
               "A1Cresult", "max_glu_serum", "other_medications",
               "repaglinide", "glimepiride", "glipizide", "glyburide", "pioglitazone", "rosiglitazone"]

# Start from the RAW (not one-hot encoded) dataframe. Encoding happens after the split.
X_raw = patient_df.drop(columns=drop_cols, errors="ignore").copy()
y = patient_df["readmitted_multiclass"].astype(int).values

# Get the patient_nbr for group splitting
groups = patient_df["patient_nbr"].values

encode_cols = [c for c in ENCODE_COLS if c in X_raw.columns]

text_cols = X_raw.select_dtypes(include=["object", "category", "string"]).columns.tolist()
binary_cols = [c for c in X_raw.columns
               if c not in text_cols and set(X_raw[c].dropna().unique()) <= {0, 1}]

X_raw[binary_cols] = X_raw[binary_cols].astype(int)
smotenc_cat_cols = text_cols + binary_cols          # treated as categories by SMOTENC
count_cols = [c for c in X_raw.columns if c not in smotenc_cat_cols]   # numeric counts
print(f"Columns to one-hot encode: {len(encode_cols)}")
print(f"SMOTENC categorical columns: {len(smotenc_cat_cols)} | numeric columns: {len(count_cols)}")
```

```python
from sklearn.preprocessing import OneHotEncoder

# This function will do One Hot Encoding to the training rows.
def fit_encoder(raw_train):
    ##Learn the one-hot categories from TRAINING rows only.
    ## drop='first' keeps the same 110-column layout as the EDA notebook
    encoder = OneHotEncoder(drop="first", handle_unknown="ignore", sparse_output=False, dtype=float)
    encoder.fit(raw_train[encode_cols])
    return encoder

# This function will apply the encoder and output the complete dataframe including th enumerical columns and encoded columns.
def apply_encoder(encoder, raw):
    ##One-hot encode rows with an encoder that was already fitted on training data.
    ##Return encoded dataframe
    dummies = encoder.transform(raw[encode_cols])
    dummies = pd.DataFrame(dummies, index=raw.index,
                           columns=encoder.get_feature_names_out(encode_cols))
    numeric = raw.drop(columns=encode_cols).astype(float)
    return pd.concat([numeric, dummies], axis=1)

# Function for resampling using SMOTENC
def smotenc_resample(raw_fit, y_fit, encoder):
    ##SMOTENC on raw training rows -> round counts back to whole numbers ->
    ## one-hot encode with the encoder fitted on the same training rows
    cat_idx = [raw_fit.columns.get_loc(c) for c in smotenc_cat_cols]
    sampler = SMOTENC(categorical_features=cat_idx, random_state=SEED)
    raw_res, y_res = sampler.fit_resample(raw_fit, y_fit)
    raw_res[count_cols] = raw_res[count_cols].round().astype(int)
    raw_res[binary_cols] = raw_res[binary_cols].astype(int)
    return apply_encoder(encoder, raw_res), np.asarray(y_res)
```

```python
## SPLIT THE DATA BY PATIENT_NBR TO AVOID DATA LEAKAGE

#  Hold-out test set, grouped by patient. We split the RAW data first.
train_idx, test_idx = next(GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=SEED).split(X_raw, y, groups))
X_raw_train, X_raw_test = X_raw.iloc[train_idx], X_raw.iloc[test_idx]
y_train, y_test = y[train_idx], y[test_idx]
groups_train = groups[train_idx]

# One-hot encoding AFTER the split: the encoder is fitted on the training data only.
encoder = fit_encoder(X_raw_train)
X_train = apply_encoder(encoder, X_raw_train)

# The test set stays raw and untouched. For the final model validation, transform it with the
# training encoder (never refit it):  X_test = apply_encoder(encoder, X_raw_test)
print(f"Train (encoded): {X_train.shape} | Test (raw, untouched): {X_raw_test.shape} | "
      f"Patients in both: {len(set(groups_train) & set(groups[test_idx]))}")
```

```python
# Define the models we will be using
MODELS = ["LogReg - Balanced", "LogReg - SMOTENC", "LightGBM - Balanced", "LightGBM - SMOTENC"]

# LightGBM Hyperparameters
LGB_PARAMS = dict(n_estimators=600, learning_rate=0.03, num_leaves=31, min_child_samples=50,
                  subsample=0.8, subsample_freq=1, colsample_bytree=0.7,
                  verbose=-1, random_state=SEED)

# This function will fit the models to the training set, it will return the probability output and model
def fit_predict(model_name, X_fit, y_fit, X_eval):
    """Fit one of the 4 models and return class probabilities.
    For SMOTENC models, X_fit / y_fit must already be resampled.
    Balanced models use class weights; SMOTENC models do not (data is already balanced)."""
    weights = "balanced" if model_name.endswith("Balanced") else None

    # This if block will run the Logistic Regression
    if model_name.startswith("LogReg"):
        scaler = StandardScaler().fit(X_fit)
        model = LogisticRegression(max_iter=2000, class_weight=weights)
        model.fit(scaler.transform(X_fit), y_fit)
        return model, model.predict_proba(scaler.transform(X_eval))

    # Else, it will run the LightGBM Classifier
    model = lgb.LGBMClassifier(**LGB_PARAMS, class_weight=weights).fit(np.asarray(X_fit), y_fit)
    return model, model.predict_proba(np.asarray(X_eval))

# This function is for output of the validation
def score(y_true, proba):
    pred = proba.argmax(axis=1)
    recall = recall_score(y_true, pred, average=None)
    return {"Macro F1": f1_score(y_true, pred, average="macro"),
            "Balanced Acc": balanced_accuracy_score(y_true, pred),
            "Macro AUC": roc_auc_score(y_true, proba, multi_class="ovr", average="macro"),
            "Recall NO": recall[0], "Recall <30": recall[1], "Recall >30": recall[2]}
```

```python
# This function prepares the training set into folds for StratifiedGroupKFold
def prepare_folds(n_folds=N_FOLDS):
    cv = StratifiedGroupKFold(n_splits=n_folds, shuffle=True, random_state=SEED)
    fold_data = []

    # This for loop goes through the raw X train set and y train set
    for fold, (fit_idx, val_idx) in enumerate(cv.split(X_raw_train, y_train, groups_train)):
        raw_fit, raw_val = X_raw_train.iloc[fit_idx], X_raw_train.iloc[val_idx]
        y_fit = y_train[fit_idx]

        # Fit the encoder on this fold's training rows only, then apply it to the validation rows
        fold_encoder = fit_encoder(raw_fit)
        X_fit = apply_encoder(fold_encoder, raw_fit)
        X_val = apply_encoder(fold_encoder, raw_val)
        X_fit_sm, y_fit_sm = smotenc_resample(raw_fit, y_fit, fold_encoder)

        # Define the dictionary for the return
        fold_data.append({
            "fold": fold + 1,
            "ranking": rank_features(X_fit, y_fit),      # This ranks the feature for the REAL Training set only
            "X_fit": X_fit, "y_fit": y_fit,
            "X_fit_sm": X_fit_sm.astype("float32"), "y_fit_sm": y_fit_sm,
            "X_val": X_val, "y_val": y_train[val_idx],
        })
        print(f"Fold {fold + 1}/{n_folds} prepared - {X_fit.shape[1]} encoded columns - "
              f"SMOTENC rows: {len(y_fit)} -> {len(y_fit_sm)}")
    return fold_data

# Run this once to prepare folds and use across all tiers
# This will take a few minutes
fold_data = prepare_folds()
```

```python
# This code block prepares the functions for building the tiers.
TOP_N = 20
GROUP_PREFIXES = [
    "discharge_disposition_", "admission_source_", "admission_type_",
    "diag_1_category_", "diag_2_category_", "diag_3_category_",
]

def build_tier_1(ranking, all_columns, top_n=TOP_N):
    """TIER 1
    Top 20 encoded columns by max one-vs-rest |r|."""
    return list(ranking.index[:top_n])

def build_tier_2(ranking, all_columns):
    """TIER 2
    Tier 1 + every remaining column of the grouped categorical features
    (discharge disposition, admission source/type, diag_1/2/3 categories)."""
    tier1 = build_tier_1(ranking, all_columns)
    group_cols = [c for c in all_columns if c.startswith(tuple(GROUP_PREFIXES))]
    return tier1 + [c for c in group_cols if c not in tier1]

def build_tier_3(ranking, all_columns):
    """TIER 3
    All encoded columns."""
    return list(all_columns)

def describe_tier(tier_builder, ranking=None):
    """Quick check of a tier on the full training set: size and columns."""
    ranking = rank_features(X_train, y_train) if ranking is None else ranking
    cols = tier_builder(ranking, X_train.columns)
    print(f"{tier_builder.__name__}: {len(cols)} columns")
    return cols
```

```python
METRICS = ["Macro F1", "Balanced Acc", "Macro AUC", "Recall <30"]

# This function is for training and validating the training set using the models
def validate_tier(tier_name, tier_builder, fold_data, models=MODELS, save_csv=True):
    """Cross-validate one tier. The tier is rebuilt inside each fold from that fold's ranking.
    Displays the summary, saves per-fold results to CSV, and returns them."""
    rows = []

    # This for loop will go through the prepare_folds dictionary for each fold
    for fd in fold_data:
        cols = tier_builder(fd["ranking"], fd["X_fit"].columns)  # Build the tiers within the fold to avoid leakage

        # Loop through each models
        for model_name in models:
            if model_name.endswith("SMOTENC"):
                _, proba = fit_predict(model_name, fd["X_fit_sm"][cols], fd["y_fit_sm"], fd["X_val"][cols])
            else:
                _, proba = fit_predict(model_name, fd["X_fit"][cols], fd["y_fit"], fd["X_val"][cols])
            rows.append({"Fold": fd["fold"], "Tier": tier_name, "Model": model_name,
                         "Columns": len(cols), **score(fd["y_val"], proba)})
    results = pd.DataFrame(rows)

    # Output results dataframe
    cols_range = f"{results['Columns'].min()}-{results['Columns'].max()}" \
        if results["Columns"].nunique() > 1 else str(results["Columns"].iloc[0])
    print(f"\n=== {tier_name} ({cols_range} columns) - {len(fold_data)}-fold CV, mean / std ===")
    display(results.groupby("Model", sort=False)[METRICS].agg(["mean", "std"]).round(3))

    # Save as CSV
    if save_csv:
        path = f"cv_results_{tier_name.lower().replace(' ', '_')}.csv"
        results.to_csv(path, index=False)
        print(f"Saved per-fold results to {path}")
    return results

def check_tier_stability(tier_builder, fold_data):
    sets = [set(tier_builder(fd["ranking"], fd["X_fit"].columns)) for fd in fold_data]
    always_in, ever_in = set.intersection(*sets), set.union(*sets)
    print(f"\n{tier_builder.__name__} stability: {len(always_in)} columns in every fold")
    print("Columns in only some folds:", sorted(ever_in - always_in) or "none")
```
